```python
def get_bot_response(user_text):
    # Convert input to lowercase and remove leading/trailing spaces
    text = user_text.lower().strip()

    # Predefined replies using if-elif statements
    if text == "hello":
        return "Hi!"
    elif text == "how are you" or text == "how are you?":
        return "I'm fine, thanks!"
    elif text == "bye":
        return "Goodbye!"
    else:
        return "I only understand 'hello', 'how are you', and 'bye'. Try again!"

def run_chatbot():
    print("--- Simple Chatbot ---")
    print("(Type 'bye' to exit the chat)\n")

    # Continuous chat loop
    while True:
        user_input = input("You: ")

        # Call the function to generate a response
        bot_reply = get_bot_response(user_input)
        print(f"Bot: {bot_reply}")

        # Exit condition to break out of the loop
        if bot_reply == "Goodbye!":
            break

if __name__ == "__main__":
    run_chatbot()
```
